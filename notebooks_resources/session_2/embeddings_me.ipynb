```python
from sentence_transformers import SentenceTransformer
```

```python
EMBEDDING_MODEL = "all-MiniLM-L6-v2"
embedder = SentenceTransformer(EMBEDDING_MODEL)
```

```python
# Example Documents
docs = [
    "Dogs are loyal and friendly domestic animals.",                  # text_1
    "Cats are independent and curious creatures.",                    # text_2
    "The Milky Way galaxy contains over 200 billion stars.",
    "Artificial intelligence is a fascinating field of study.",
    "Python is a popular programming language.",
    "My cat is lucy"           
    ]
```

```python
embeddings = embedder.encode(docs)
```

```python
embeddings.shape
```

```python
custom_query = "What is the Milky Way?"
```

```python
query_embedding = embedder.encode(custom_query)
```

```python
scores = embedder.similarity(query_embedding, embeddings)
```
